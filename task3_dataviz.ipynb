```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("MyAnimeList_TopAnime.csv")

df.head(3)
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="Score", kde=True, color="indigo", bins=10)

plt.title("Score Distribution Among the Top 50 Anime", fontsize=16, fontweight='bold')
plt.xlabel("MyAnimeList Score", fontsize=12)
plt.ylabel("Number of Anime", fontsize=12)

plt.show()
```

```python
#We want to see the relationship between Rank and Score
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x="Rank", y="Score", color="crimson", s=100)

plt.title("Score Decay: The Gap Between #1 and #50", fontsize=16, fontweight='bold')
plt.xlabel("Anime Rank", fontsize=12)
plt.ylabel("MAL Score", fontsize=12)

plt.show()
```

```python
top_10 = df.head(10)
plt.figure(figsize=(12, 6))

sns.barplot(data=top_10, x="Score", y="Title", hue="Title", palette="magma", legend=False)

plt.title("Top 10 Anime by Score (Seaborn Edition)", fontsize=16, fontweight='bold')
plt.xlabel("MAL Score", fontsize=12)
plt.ylabel("")

plt.xlim(8.9, 9.3)

plt.show()
```
