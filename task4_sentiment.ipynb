```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob

#testing on a sentence
test_sentence = "Frieren is an absolute masterpiece, the animation is gorgeous!"
blob = TextBlob(test_sentence)

score= blob.sentiment.polarity
print(f"The emotional score is: {score}")
```

```python
#custom data of anime reviews
review_data = {
    'Anime': ['Frieren', 'Show B', 'Show C', 'FMAB', 'Show E'],
    'ReviewText': [
        "Absolutely beautiful story with stunning animation.",
        "I hated the new season, it was boring and the pacing was terrible.",
        "It was okay, nothing special but not bad either.",
        "The greatest story ever told, a true masterpiece.",
        "The ending was incredibly disappointing and ruined the show."

    ]
}

df_reviews = pd.DataFrame(review_data)

def analyze_emotion(text):
    score = TextBlob(text).sentiment.polarity
    if score > 0.1:
        return 'Positive'
    elif score < -0.1:
        return 'Negative'
    else:
        return 'Neutral'

df_reviews['Sentiment_result'] = df_reviews['ReviewText'].apply(analyze_emotion)

df_reviews
```

```python
# creating a bar chart counting the sentiments
plt.figure(figsize=(8, 5))
sns.countplot(data=df_reviews, x="Sentiment_result", hue="Sentiment_result", palette=['#4CAF50', '#F44336'], legend=False)

plt.title("Anime Reviews: Sentiment Analysis Breakdown", fontweight='bold')
plt.xlabel("Detected Emotion")
plt.ylabel("Number of Reviews")

plt.show()
```
